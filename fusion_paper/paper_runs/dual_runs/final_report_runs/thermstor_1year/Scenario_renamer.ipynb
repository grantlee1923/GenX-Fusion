Just a code to change old folders not valid anymore into another thing

```python
import sys
import os
import shutil

current_path = os.getcwd()
```

```python
folder_to_remove_generic_name = "FlexibleDemand100MW4hr_Cost_3000.0_EmissLevel_4.0_gCO2perkWh"

user_confirmation = str(input(f"You're about to DELETE ALL {folder_to_remove_generic_name}...Are you sure? (Y/N)"))
if user_confirmation.capitalize() != "Y":
    raise ValueError("Confirmation not given")
```

```python
if user_confirmation.capitalize() == "Y":
    print("Confirmation granted")
    for scenario in os.listdir(current_path+"\Scenarios"):
        print(f"Updating scenario: {scenario}")
        if scenario != "Scenario_00":
            local_folder_name_to_remove = f"{current_path}\Scenarios\{scenario}\Results\{folder_to_remove_generic_name}"
            try:
                shutil.rmtree(local_folder_name_to_remove)
                print(f"Folder removed successfully.")
            except OSError as e:
                print(f"Error removing folder: {e}")
```
